#ALJABAR LINEAR

```python
import numpy as np

# Vectors
a = np.array([1, 2])
b = np.array([3, 4])
print("a =",  a)
print("b =", b)
```

```python
# Dot product
dot = np.dot(a, b)  # 1*3 + 2*4 = 11
print("hasil operasi dot product =", dot)
```

```python
# Matrix multiplication
X = np.array([[1, 2], [3, 4]])
W = np.array([[0.5, 0], [0, 0.5]])

Y = X @ W  # element-wise dot
print("hasil matrix  =", X,  "di kali", W, "=",  Y)
```

#KALKULUS & OPTIMISASI

```python
# Sample data
X = np.array([[1], [2], [3]])
y = np.array([2, 4, 6])
w = 0.0  # initial weight
lr = 0.01  # learning rate

# Gradient descent
for i in range(100):
    y_pred = X * w
    loss = ((y_pred - y)**2).mean()
    grad = 2 * (y_pred - y).mean() * X.mean()
    w -= lr * grad
    print(f"Iterasi {i+1}: loss = {loss}, w = {w}")
```

```python
print(f" bobot akhir setelah pelatihan w =   {w:.6f}")
```

#Statistik Probability

```python
from sklearn.naive_bayes import BernoulliNB

# X: binary features, y: labels
X = [[1,0,1],[0,1,1],[1,1,0]]
y = [1, 0, 1]

model = BernoulliNB()
model.fit(X, y)
pred = model.predict([[1,0,0]])
print("Prediction:", pred)
```

#Exponential and SoftMax

```python
import numpy as np

# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

print("nilai sigmoid pada x 0 = ", sigmoid(0))  # 0.5

# Softmax
def softmax(z):
    e_z = np.exp(z - np.max(z))
    return e_z / e_z.sum()

scores = np.array([2.0, 1.0, 0.1])
print("Score Sotfmax = ",softmax(scores))  # probabilities sum to 1
```

#EigenValues and PCA

```python
from sklearn.decomposition import PCA

X = np.array([[2.5, 2.4],
              [0.5, 0.7],
              [2.2, 2.9],
              [1.9, 2.2]])

pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X)
print(X_reduced)
```
